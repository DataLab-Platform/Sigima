# Sigima Simple Example

This notebook demonstrates some features of **Sigma**, the signal and image processing library that powers [DataLab](https://datalab-platform.com).

While Sigima’s core computing features are now stable, its Jupyter-related capabilities are still at an early stage of development.

```python
import numpy as np
import sigima.objects
import sigima.proc.image

data = np.random.normal(100, 30, (100, 100))  # Prepare test image data

img = sigima.objects.create_image("Noisy", data)  # Create the image object
img.roi = sigima.objects.create_image_roi("circle", [30, 30, 20])  # Define ROI
result = sigima.proc.image.gaussian_filter(img, sigma=5.0)  # Apply Gaussian filter
result
```

```python
from sigima import viz
viz.view_images([img, result])
```
